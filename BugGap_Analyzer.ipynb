```python
# ============================================================
# RESPONSEEVAL AI
# MISSED BUGS AND INCORRECT SUGGESTIONS
# ============================================================

import pandas as pd

data = [
    {
        "id": 1,
        "code": "result = 10 / 0",
        "expected_bug": "ZeroDivisionError",
        "detected_bug": "ZeroDivisionError",
        "suggestion": "Check whether the denominator is zero before division.",
        "correct_suggestion": True
    },

    {
        "id": 2,
        "code": "numbers = [10, 20, 30]\nprint(numbers[5])",
        "expected_bug": "IndexError",
        "detected_bug": "IndexError",
        "suggestion": "Check that the index is within the valid range.",
        "correct_suggestion": True
    },

    {
        "id": 3,
        "code": "name = None\nprint(name.upper())",
        "expected_bug": "AttributeError",
        "detected_bug": "No Issue",
        "suggestion": "No changes required.",
        "correct_suggestion": False
    },

    {
        "id": 4,
        "code": "age = int('abc')",
        "expected_bug": "ValueError",
        "detected_bug": "TypeError",
        "suggestion": "Convert the value to a floating-point number.",
        "correct_suggestion": False
    },

    {
        "id": 5,
        "code": "numbers = []\naverage = sum(numbers) / len(numbers)",
        "expected_bug": "ZeroDivisionError",
        "detected_bug": "ZeroDivisionError",
        "suggestion": "Check whether the list is empty before calculating the average.",
        "correct_suggestion": True
    },

    {
        "id": 6,
        "code": "data = {'name': 'John'}\nprint(data['age'])",
        "expected_bug": "KeyError",
        "detected_bug": "No Issue",
        "suggestion": "Use data.get('age') or check whether the key exists.",
        "correct_suggestion": False
    }
]

df = pd.DataFrame(data)

print("Dataset created successfully.")
print("Total test cases:", len(df))

df
```

```python
# ============================================================
# MISSED BUG DETECTION
# ============================================================

df["bug_missed"] = (
    df["expected_bug"] != df["detected_bug"]
)

missed_bugs = df[df["bug_missed"] == True]

print("=" * 60)
print("                MISSED BUGS")
print("=" * 60)

if len(missed_bugs) == 0:
    print("No missed bugs found.")
else:
    for _, row in missed_bugs.iterrows():
        print(f"\nTest Case : {row['id']}")
        print(f"Expected  : {row['expected_bug']}")
        print(f"Detected  : {row['detected_bug']}")
        print(f"Code      : {row['code']}")

print("\nTotal Missed Bugs:", len(missed_bugs))
```

```python
# ============================================================
# INCORRECT SUGGESTION DETECTION
# ============================================================

df["incorrect_suggestion"] = (
    df["correct_suggestion"] == False
)

incorrect_suggestions = df[
    df["incorrect_suggestion"] == True
]

print("=" * 60)
print("             INCORRECT SUGGESTIONS")
print("=" * 60)

if len(incorrect_suggestions) == 0:
    print("No incorrect suggestions found.")
else:
    for _, row in incorrect_suggestions.iterrows():

        print(f"\nTest Case : {row['id']}")
        print(f"Expected Bug : {row['expected_bug']}")
        print(f"Detected Bug : {row['detected_bug']}")
        print(f"Suggestion   : {row['suggestion']}")

print("\nTotal Incorrect Suggestions:",
      len(incorrect_suggestions))
```

```python
# ============================================================
# BUG DETECTION RATE
# ============================================================

total_bugs = len(df)

correctly_detected = len(
    df[df["expected_bug"] == df["detected_bug"]]
)

detection_rate = (
    correctly_detected / total_bugs
) * 100

print("=" * 60)
print("              BUG DETECTION RATE")
print("=" * 60)

print("Total Bugs          :", total_bugs)
print("Correctly Detected  :", correctly_detected)
print("Missed Bugs         :", len(missed_bugs))
print(f"Detection Rate      : {detection_rate:.2f}%")
```

```python
# ============================================================
# SUGGESTION ACCURACY
# ============================================================

correct_suggestions = len(
    df[df["correct_suggestion"] == True]
)

total_suggestions = len(df)

suggestion_accuracy = (
    correct_suggestions / total_suggestions
) * 100

print("=" * 60)
print("             SUGGESTION ACCURACY")
print("=" * 60)

print("Total Suggestions   :", total_suggestions)
print("Correct Suggestions :", correct_suggestions)
print("Incorrect Suggestions:",
      len(incorrect_suggestions))

print(
    f"Suggestion Accuracy : "
    f"{suggestion_accuracy:.2f}%"
)
```

```python
# ============================================================
# OVERALL EVALUATION
# ============================================================

overall_score = (
    detection_rate * 0.5
    + suggestion_accuracy * 0.5
)

if overall_score >= 90:
    performance = "Excellent"
elif overall_score >= 75:
    performance = "Good"
elif overall_score >= 60:
    performance = "Moderate"
else:
    performance = "Needs Improvement"

print("=" * 60)
print("             RESPONSEEVAL AI")
print("       BUG & SUGGESTION EVALUATION")
print("=" * 60)

print(f"Bug Detection Rate     : {detection_rate:.2f}%")
print(f"Suggestion Accuracy    : {suggestion_accuracy:.2f}%")
print(f"Missed Bugs            : {len(missed_bugs)}")
print(f"Incorrect Suggestions  : {len(incorrect_suggestions)}")
print(f"Overall Score          : {overall_score:.2f}%")
print(f"Performance            : {performance}")

print("=" * 60)
```
